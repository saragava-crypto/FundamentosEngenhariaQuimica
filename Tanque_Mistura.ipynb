```python
from dataclasses import dataclass

@dataclass
class Corrente:
  """
  Vazão: vazão da corrente em kg/h
  Fração: fração mássica em kg/kg
  """
  vazao: float
  fracao: float
```

```python
def misturador(c1: Corrente, c2: Corrente):

#Primeira Corrente
  Fa = c1. vazao
  xa = c1.fracao

#Segunda Corrente
  Fb = c2. vazao
  xb = c2.fracao

#Vazão Total
  F = Fa + Fb

#Fração mássica Total
  x = (Fa*xa + Fb*xb)/F

  return F,x

```
